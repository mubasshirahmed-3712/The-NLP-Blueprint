<h2 style="text-align:center;">Word Embeddings (Word2Vec, CBOW & Skip-gram)</h2>


### 🔹 Introduction

- **Word embeddings** are dense vector representations of words that capture **semantic meaning** and **relationships**.  
- Unlike **one-hot encoding**, embeddings place similar words close together in a high-dimensional space.  
- Example: *king - man + woman ≈ queen*.  


### 🔹 Word2Vec Basics

- **Word2Vec** (developed by Google, 2013) learns embeddings from text.  
- Two training architectures:

1. **CBOW (Continuous Bag of Words):** Predicts a word from its context.  
2. **Skip-gram:** Predicts context words from a single target word.  

| Feature      | CBOW                          | Skip-gram                     |
|--------------|-------------------------------|-------------------------------|
| Input        | Context words                 | Target word                   |
| Output       | Target word                   | Context words                 |
| Best For     | Smaller datasets              | Larger datasets               |
| Strength     | Captures global meaning       | Captures fine-grained details |


### 🔹 Advantages
- Efficient on large corpora  
- Captures semantic similarity better than BoW/TF-IDF  
- Pre-trained models available (e.g., Google News Word2Vec)  

### 🔹 Limitations
- **Static embeddings:** one vector per word (no context awareness).  
- Struggles with **polysemy** (e.g., *bank* 🏦 vs *river bank* 🌊).  
- Cannot handle **out-of-vocabulary (OOV)** words.  


```python
import gensim
from gensim.models import Word2Vec
import warnings
warnings.filterwarnings("ignore", category=UserWarning, module='gensim')

# Sample training sentences
sentences = [
    ["I", "love", "natural", "language", "processing"],
    ["Word2Vec", "is", "a", "great", "tool"],
    ["Machine", "learning", "is", "fun"],
]

# Train Word2Vec model (Skip-gram: sg=1)
model = Word2Vec(sentences, vector_size=100, window=5, min_count=1, sg=1)

# ➤ Get the vector for a word
print("Vector for 'language':\n", model.wv['language'])

# ➤ Find most similar words
print("\nWords similar to 'language':")
print(model.wv.most_similar('language', topn=5))
```

### 🔹 CBOW vs Skip-gram

**1. Continuous Bag of Words (CBOW):**  
- Input: Context words → Output: Target word  
- Example: In "The cat sits on the mat", context = ["The", "cat", "on", "the", "mat"], target = "sits".  
- Works well with **smaller datasets**, captures global meaning.  

**2. Skip-gram:**  
- Input: Target word → Output: Context words  
- Example: Target = "sits", predicts ["The", "cat", "on", "the", "mat"].  
- Works well with **larger datasets**, captures detailed semantic relations.  


```python
# Training Data
sentences = [
    ["I", "love", "natural", "language", "processing"],
    ["Word2Vec", "is", "a", "great", "tool"],
    ["Machine", "learning", "is", "fun"],
    ["Natural", "language", "processing", "is", "awesome"]
]

# CBOW Model (sg=0)
cbow_model = Word2Vec(sentences, vector_size=100, window=2, min_count=1, sg=0)

# Skip-gram Model (sg=1)
skipgram_model = Word2Vec(sentences, vector_size=100, window=2, min_count=1, sg=1)

word = "language"

# ➤ Get vectors
print(f"CBOW Vector for '{word}':\n", cbow_model.wv[word])
print(f"\nSkip-gram Vector for '{word}':\n", skipgram_model.wv[word])

# ➤ Find similar words
print(f"\nCBOW - Words similar to '{word}':")
print(cbow_model.wv.most_similar(word, topn=5))

print(f"\nSkip-gram - Words similar to '{word}':")
print(skipgram_model.wv.most_similar(word, topn=5))
```

### ✅ Summary

- **Word embeddings** help capture semantic meaning in vector form.  
- **CBOW**: Faster, works well for small data.  
- **Skip-gram**: Slower, but better for large datasets and semantic detail.  
- Word2Vec paved the way for advanced embeddings:
  - **GloVe** (Global Vectors)  
  - **FastText** (handles subwords & OOV words)  
  - **Contextual embeddings**: BERT, ELMo, GPT.
