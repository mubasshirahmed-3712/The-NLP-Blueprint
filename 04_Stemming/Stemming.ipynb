<h2 style="text-align:center;">🌱 Stemming in NLP</h2>

---

### 🔹 What is Stemming?
Stemming is the process of **reducing words to their base/root form** by chopping off suffixes/prefixes.

👉 Example:  
- "playing" → "play"  
- "studies" → "studi" (❌ not always meaningful)  

---

### 🔹 Why Use Stemming?
- Reduces vocabulary size.  
- Groups similar words together.  
- Useful in tasks like search engines.  

---

### 🔹 Limitations
- Sometimes over-stems → produces non-dictionary words.  
- Example: *"better"* → remains *"better"* (not reduced properly).  

---


```python
# ➤ Import libraries
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

# Initialize stemmer
ps = PorterStemmer()

# Sample words
words = ["playing", "played", "plays", "studies", "studying", "better"]

print("Original → Stemmed")
for w in words:
    print(w, "→", ps.stem(w))
```

---

### ✅ Summary
- **Stemming** = chopping suffixes to get root word.  
- Fast, but sometimes produces strange roots (studi, etc.).
