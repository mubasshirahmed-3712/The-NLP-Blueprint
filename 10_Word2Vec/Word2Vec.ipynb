<h2 style="text-align:center;">🧠 Word2Vec</h2>

---

### 🔹 What is Word2Vec?
Word2Vec is a **neural network-based embedding technique** that represents words as dense vectors (not just counts).  

👉 Key idea: **Words with similar meaning have similar vectors.**  
Example:  
- "king" - "man" + "woman" ≈ "queen"  

---

### 🔹 Models in Word2Vec
1. **CBOW (Continuous Bag of Words):** Predicts a word from context.  
2. **Skip-gram:** Predicts context words from a target word.  

---

### 🔹 Why Better?
- Captures **semantic meaning**.  
- Preserves word relationships.  

---


```python
# ➤ Install gensim if needed
# ! pip install gensim
```

```python


from gensim.models import Word2Vec

# Sample corpus
sentences = [
    ["i", "love", "nlp"],
    ["i", "love", "ai"],
    ["nlp", "and", "ai", "are", "amazing"]
]

# Train Word2Vec model
model = Word2Vec(sentences, vector_size=50, window=3, min_count=1, workers=4)

# Get vector for a word
print("Vector for 'nlp':\n", model.wv["nlp"])

# Most similar words
print("\nMost similar to 'ai':\n", model.wv.most_similar("ai"))
```

---

### ✅ Summary
- **Word2Vec** = word embeddings using neural networks.  
- Captures meaning & relationships between words.  
- Two models: **CBOW** & **Skip-gram**.
