<h2 style="text-align:center;">Chunking in NLP</h2>

---

### 📌 Introduction
- **Chunking** in Natural Language Processing (NLP) is the process of **grouping words into meaningful phrases** (called *chunks*).  
- These chunks typically correspond to syntactic units like **Noun Phrases (NP)**, **Verb Phrases (VP)**, or **Prepositional Phrases (PP)**.  
- Unlike tokenization (splitting text into words) and POS tagging (assigning parts of speech), chunking focuses on **higher-level structures** in sentences.  

💡 Example:  
For the sentence → *"The quick brown fox jumps over the lazy dog"*  

- **NP (Noun Phrase):** "The quick brown fox"  
- **VP (Verb Phrase):** "jumps"  
- **PP (Prepositional Phrase):** "over the lazy dog"  

---


```python
# ➤ Importing required libraries
import nltk
from nltk import pos_tag, word_tokenize, RegexpParser


# Download necessary NLTK datasets (only once)
# nltk.download('punkt')
# nltk.download('averaged_perceptron_tagger')
```

### 🧠 Step 1: Tokenization
- First, we **tokenize** the input text into individual words.
- Tokenization helps break the raw text into manageable units.


```python
# ➤ Sample text
text = "The quick brown fox jumps over the lazy dog."

# ➤ Tokenizing text
tokens = word_tokenize(text)
print("Tokens:", tokens)
```

### 🧠 Step 2: POS Tagging
- Each token is assigned a **Part-of-Speech (POS)** tag.
- POS tags help the chunker identify which tokens can be grouped together.


```python
# ➤ POS tagging on tokens
tagged_tokens = pos_tag(tokens)
print("POS Tags:", tagged_tokens)
```

### 🧠 Step 3: Defining Chunk Grammar
- Chunking uses a **regular expression grammar** to define how words should be grouped.  
- Example rules:  
  - `NP: {<DT>?<JJ>*<NN>}` → A Noun Phrase may start with a determiner, followed by adjectives, ending with a noun.  
  - `VP: {<VB.*><NP|PP>*}` → A Verb Phrase starts with a verb, followed by NPs or PPs.  
  - `PP: {<IN><NP>}` → A Prepositional Phrase starts with a preposition followed by an NP.


```python
# ➤ Define a chunk grammar
chunk_grammar = r"""
  NP: {<DT>?<JJ>*<NN>}   # Noun Phrase
  VP: {<VB.*><NP|PP>*}   # Verb Phrase
  PP: {<IN><NP>}         # Prepositional Phrase
"""

# ➤ Create a chunk parser
chunk_parser = RegexpParser(chunk_grammar)
```

### 🧠 Step 4: Chunk Parsing
- Apply the defined grammar to POS-tagged tokens.
- This will group tokens into **NP, VP, and PP chunks**.


```python
# ➤ Parse the tagged tokens
chunked = chunk_parser.parse(tagged_tokens)

# ➤ Print the chunked structure
print(chunked)
```

### 🧠 Step 5: Visualizing Chunks
- We can visualize the chunks using `chunked.draw()`.
- A tree diagram will show **how tokens are grouped into phrases**.


```python
# ➤ Visualize chunk tree (will open a window in Jupyter/Colab)
chunked.draw()
```

<h2 style="text-align:center;">📌 Summary</h2>

- **Chunking** groups words into **syntactic phrases** (NP, VP, PP).  
- It uses **tokenization + POS tagging + grammar rules**.  
- Useful in:
  - **Information Extraction** (finding entities, relations)  
  - **Text Parsing** (syntactic structure)  
  - **Question Answering & Sentiment Analysis** (extracting meaningful parts of text)
