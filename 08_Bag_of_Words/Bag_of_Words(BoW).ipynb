<h2 style="text-align:center;">👜 Bag of Words (BoW)</h2>

---

### 🔹 What is BoW?
Bag of Words is a text representation technique where:
- Each document is represented as a **vector of word counts**.
- Ignores grammar and order of words.

👉 Example:  
Sentences:  
1. "I love NLP"  
2. "I love AI"  

BoW → vocabulary: ["I", "love", "NLP", "AI"]  
- Doc1 → [1, 1, 1, 0]  
- Doc2 → [1, 1, 0, 1]  

---

### 🔹 Use Cases
- Simple text classification.  
- Spam detection, sentiment analysis.  

---


```python
# ➤ Import libraries
from sklearn.feature_extraction.text import CountVectorizer

# Sample documents
docs = [
    "I love NLP",
    "I love AI",
    "AI and NLP are amazing"
]

# Initialize BoW
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(docs)

# Vocabulary
print("Vocabulary:\n", vectorizer.get_feature_names_out())

# Document-term matrix
print("\nBoW Representation:\n", X.toarray())
```

---

### ✅ Summary
- **BoW** = simple word count representation.  
- Easy, but ignores order and meaning.
