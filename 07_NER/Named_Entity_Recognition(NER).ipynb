<h2 style="text-align:center;">🏷️ Named Entity Recognition (NER)</h2>

---

### 🔹 What is NER?
NER identifies **real-world entities** (names, places, dates, organizations) in text.

👉 Example:  
- "Elon Musk founded SpaceX in 2002"  
➡ [(Elon Musk, PERSON), (SpaceX, ORG), (2002, DATE)]  

---

### 🔹 Why Use NER?
- Extracts meaningful information.  
- Widely used in resumes, finance, medical, and news extraction.  

---


```python
# ➤ Import libraries
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk

# nltk.download("maxent_ne_chunker")
# nltk.download("words")

# Sample text
text = "Elon Musk founded SpaceX in 2002."

# Tokenize + POS Tag + NER
tokens = word_tokenize(text)
tags = pos_tag(tokens)
entities = ne_chunk(tags)

print("Named Entities:\n", entities)
```

---

### ✅ Summary
- **NER** detects persons, organizations, locations, dates.  
- Example: "Elon Musk" → PERSON, "2002" → DATE.  
- Important in information extraction.
