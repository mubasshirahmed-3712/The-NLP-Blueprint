<h2 style="text-align:center;">⚡ spaCy Basics – A One Stop Guide</h2>


### 📌 Introduction
- **spaCy** is a popular industrial-strength **NLP library in Python**.  
- It is designed for **fast, efficient, and production-ready NLP pipelines**.  
- Supports multiple languages, pre-trained models, and integration with deep learning frameworks.  

👉 spaCy is widely used for:
- Tokenization  
- Lemmatization  
- POS tagging  
- Named Entity Recognition (NER)  
- Dependency Parsing  
- Sentence segmentation  
- Pretrained word vectors  


```python
# ➤ Install spaCy (only needed once in environment)
# !pip install spacy

# ➤ Download small English model
# !python -m spacy download en_core_web_sm
```

```python
import spacy

# Load small English model
nlp = spacy.load("en_core_web_sm")

# Sample text
doc = nlp("Apple is looking at buying U.K. startup for $1 billion in 2024.")

doc
```

### 🔹 Tokenization
- Splitting raw text into meaningful units (**tokens**).  
- spaCy handles punctuation, spaces, and special symbols efficiently.  


```python
for token in doc:
    print(token.text)
```

### 🔹 Token Attributes
Each token in spaCy has multiple attributes:
- `token.text` → actual word  
- `token.lemma_` → lemma/root form  
- `token.pos_` → part of speech (universal)  
- `token.tag_` → detailed POS tag  
- `token.dep_` → dependency relation  
- `token.is_stop` → stopword or not  
- `token.shape_` → word shape (e.g., Xxxxx, dddd)  


```python
for token in doc:
    print(token.text, token.lemma_, token.pos_, token.tag_, token.dep_, token.shape_, token.is_stop)
```

### 🔹 Sentence Segmentation
spaCy automatically splits text into sentences using its dependency parser.  


```python
text = "I love NLP. spaCy makes it so easy. It is fast and powerful."
doc = nlp(text)

for sent in doc.sents:
    print(sent)
```

### 🔹 Stopwords
- Stopwords = common words (is, the, and, etc.) often removed in preprocessing.  
- spaCy provides built-in stopword list.  


```python
from spacy.lang.en.stop_words import STOP_WORDS

print("Number of stopwords:", len(STOP_WORDS))
print("First 20 stopwords:", list(STOP_WORDS)[:20])
```

### 🔹 Lemmatization
- Converts words to their **base/dictionary form**.  
- Helps reduce variations of a word.  


```python
doc = nlp("The striped cats are better than the striped dogs.")
for token in doc:
    print(token.text, "->", token.lemma_)
```

### 🔹 Part-of-Speech (POS) Tagging
- Assigns grammatical category to each word (NOUN, VERB, ADJ, etc.).  


```python
doc = nlp("The quick brown fox jumps over the lazy dog.")
for token in doc:
    print(token.text, "->", token.pos_)
```

### 🔹 Named Entity Recognition (NER)
- Identifies **real-world entities** (names, places, money, dates, etc.).  


```python
doc = nlp("Apple is looking at buying U.K. startup for $1 billion in 2024.")

for ent in doc.ents:
    print(ent.text, ent.label_)
```

### 🔹 Dependency Parsing
- Analyzes **grammatical structure** of sentences.  
- Shows relationships between words (subject, object, modifiers).  


```python
for token in doc:
    print(token.text, "->", token.dep_, "->", token.head.text)
```

### 🔹 Noun Chunks
- Base noun phrases detected automatically by spaCy.  


```python
doc = nlp("The quick brown fox jumped over the lazy dog.")
for chunk in doc.noun_chunks:
    print(chunk.text)
```

### 🔹 Word Vectors
- spaCy **medium & large models** include pretrained word vectors.  
- Can be used to calculate **similarity** between words/sentences.  


```python
# Load medium model (if available)
# !python -m spacy download en_core_web_md
# nlp = spacy.load("en_core_web_md")

doc = nlp("I love data science and machine learning.")
for token in doc:
    print(token.text, token.has_vector, token.vector.shape)

# Similarity between words
doc2 = nlp("king queen man woman")
print(doc2[0].similarity(doc2[1]))  # king ~ queen
print(doc2[0].similarity(doc2[2]))  # king ~ man
```

### 🔹 spaCy Pipelines
- spaCy uses **pipelines** (sequence of components).  
- Components: Tokenizer → Tagger → Parser → NER → Custom.  


```python
print(nlp.pipe_names)
```

<h2 style="text-align:center;">✅ Summary of spaCy Basics</h2>

- **Tokenization**: Splitting text into words  
- **Attributes**: lemma, POS, dependency, stopwords  
- **Sentence Segmentation**: Auto detection of sentences  
- **Stopwords**: Built-in list  
- **Lemmatization**: Reducing to base form  
- **POS Tagging**: Grammatical category  
- **NER**: Detects names, dates, money, etc.  
- **Dependency Parsing**: Grammar relationships  
- **Noun Chunks**: Easy phrase detection  
- **Word Vectors**: Semantic similarity (medium/large models)  
- **Pipeline**: Components used in processing
