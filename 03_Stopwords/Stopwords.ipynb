<h2 style="text-align:center;">🛑 Stopwords in NLP</h2>

---

### 🔹 What are Stopwords?
Stopwords are **common words** in a language that carry little to no meaning and are usually removed during preprocessing.

👉 Examples:  
- English: *is, am, the, a, an, in, of, and*  
- Hindi: *का, की, के, यह, है*  

---

### 🔹 Why Remove Stopwords?
- They occur very frequently and **don’t add much value** to analysis.  
- Removing them reduces noise and improves efficiency.  

---

### 🔹 When *Not* to Remove Stopwords?
- Some tasks like **machine translation** or **text generation** may require them.  
- Example: Removing "not" changes the meaning completely!  
  - "I am **not** happy" → removing *not* flips sentiment.  

---


```python
# ➤ Import libraries
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python


# Download stopwords if not already installed
# nltk.download("stopwords")
# nltk.download("punkt")

# Sample text
text = "This is an example showing how stopword removal works in Natural Language Processing."
print("Original Text:\n", text)
```

```python
# ➤ Word Tokenization
words = word_tokenize(text)
print("Word Tokens:\n", words)
```

```python
# ➤ Remove Stopwords
stop_words = set(stopwords.words("english"))

filtered_words = [w for w in words if w.lower() not in stop_words]
print("After Stopword Removal:\n", filtered_words)
```

```python
# ➤ Show few English stopwords
print("Some English Stopwords:\n", list(stop_words)[:30])
```

---

### ✅ Summary
- **Stopwords** = common words (is, am, the, etc.) usually removed in NLP tasks.  
- Helps reduce noise and dimensionality.  
- But be careful: removing words like **not** can change meaning drastically.
