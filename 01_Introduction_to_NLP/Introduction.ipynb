### 🔹 What is NLP?
Natural Language Processing (**NLP**) is a field of Artificial Intelligence that enables computers to understand, interpret, and generate human language.

👉 Simply, NLP is the **bridge between humans and machines**.  
It helps machines make sense of spoken or written language.

---

### 🔹 Types of NLP Approaches
1. **Rule-based NLP**  
   - Early systems used hand-crafted grammar rules.  
   - Example: if word ends with "ing" → verb.  
   - ❌ Very rigid, poor with slang & typos.

2. **Statistical NLP**  
   - Uses probabilities and word frequencies.  
   - Example: "I ___ to school" → model predicts "go" with highest probability.

3. **Machine Learning–based NLP**  
   - Algorithms learn from data (classification, clustering, etc.).  
   - Example: Sentiment Analysis (positive vs negative).

4. **Deep Learning NLP (Modern)**  
   - Neural networks, embeddings, transformers.  
   - Example: ChatGPT, BERT, Google Translate.

---

### 🔹 Real-Life Applications
- Search Engines (Google, Bing)  
- Chatbots & Virtual Assistants (Alexa, Siri)  
- Sentiment Analysis (Twitter, Reviews)  
- Machine Translation (English ↔ Hindi)  
- Spam Detection in Emails  
- Healthcare & Finance Document Analysis  

---

### 🔹 Corpora in NLP
**Corpus** = a large collection of text documents used for training NLP models.  

👉 Examples in `nltk` library:  
- **Brown Corpus** – variety of genres.  
- **Gutenberg Corpus** – classical literature.  
- **Stopwords Corpus** – common filler words.  

---



```python
# ➤ Import libraries
import os
import nltk

# Uncomment this line if running for the first time to download resources
# nltk.download()

# List available corpora in NLTK
print(os.listdir(nltk.data.find("corpora")))
```

```python
# ➤ Example: Explore Brown Corpus
from nltk.corpus import brown

print("Available Categories:", brown.categories()[:5])  # show first 5 categories
print("\nSample Words:", brown.words()[:20])  # first 20 words
```

```python
# ➤ Example: Explore Gutenberg Corpus
from nltk.corpus import gutenberg

print("Available Files:", gutenberg.fileids())
print("\nFirst 100 words of Austen's Emma:")
print(gutenberg.words('austen-emma.txt')[:100])
```
