<h2 style="text-align:center;">📊 TF-IDF (Term Frequency - Inverse Document Frequency)</h2>

---

### 🔹 What is TF-IDF?
TF-IDF assigns **weights** to words based on importance in a document.  
- **TF (Term Frequency):** How often a word appears in a doc.  
- **IDF (Inverse Document Frequency):** How rare a word is across all docs.  

👉 Formula:  
`TF-IDF = TF × IDF`

---

### 🔹 Why TF-IDF over BoW?
- Removes bias of frequent/common words (like "the", "is").  
- Highlights rare but important words.  

---


```python
# ➤ Import libraries
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample documents
docs = [
    "I love NLP",
    "I love AI",
    "AI and NLP are amazing"
]

# Initialize TF-IDF
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)

# Vocabulary
print("Vocabulary:\n", vectorizer.get_feature_names_out())

# TF-IDF matrix
print("\nTF-IDF Representation:\n", X.toarray())
```

---

### ✅ Summary
- **TF-IDF** gives weight to words → rare words get higher scores.  
- Better than BoW for feature extraction.
