<h2 style="text-align:center;">✂️ Tokenization in NLP</h2>

---

### 🔹 What is Tokenization?
**Tokenization** is the process of splitting text into smaller units called **tokens**.  
Tokens can be **words**, **sentences**, or even **subwords**.

👉 Example:  
`"I love Natural Language Processing"`  
➡ `["I", "love", "Natural", "Language", "Processing"]`

---

### 🔹 Why Tokenization?
- Breaks raw text into analyzable pieces.  
- First step in almost every NLP pipeline.  
- Required before feeding text into ML/DL models.  

---

### 🔹 Types of Tokenization
1. **Word Tokenization** → Splits text into words.  
2. **Sentence Tokenization** → Splits text into sentences.  
3. **Character Tokenization** → Splits text into characters.  
4. **Subword Tokenization (Modern NLP)** → Splits into meaningful chunks (used in BERT, GPT).

---


```python
# ➤ Import libraries
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

# Download tokenizers if not already installed
# nltk.download("punkt")

# Sample text
text = "Natural Language Processing (NLP) is fun! Let's explore tokenization."
print("Original Text:\n", text)
```

```python
# ➤ Word Tokenization
words = word_tokenize(text)
print("Word Tokens:\n", words)
```

```python
# ➤ Sentence Tokenization
sentences = sent_tokenize(text)
print("Sentence Tokens:\n", sentences)
```

```python
# ➤ Character Tokenization (custom)
chars = list(text)
print("Character Tokens:\n", chars[:50])  # show first 50 chars
```

---

### ✅ Summary
- **Tokenization** splits text into smaller units (words, sentences, chars).  
- `nltk.tokenize` provides ready functions like `word_tokenize()` and `sent_tokenize()`.  
- Tokenization is the **first essential preprocessing step** in NLP.
